### Nominal Max Clique

```python
class solution:
    
    def __init__(self, G, lb):
        self.G = G
        self.max = lb
    
    def Clique(self, G, U, size):

        if len(U) == 0:
            if size > self.max:
                self.max = size
            return

        while len(U) > 0:
            if size + len(U) <= self.max:
                return

            u = U.pop()
            N_prime_of_u = set([w for w in G.neighbors(u) if G.degree[w] >= self.max])
            self.Clique(G, U.intersection(N_prime_of_u), size + 1)
            
    def MaxClique(self):
        
        for i, v_i in enumerate(range(self.G.number_of_nodes())):
            if self.G.degree[v_i] >= self.max:
                U = set()
                for j, v_j in enumerate(self.G.neighbors(v_i)):
                    if j > i:
                        if self.G.degree[v_j] >= self.max:
                            U.add(v_j)
                self.Clique(G, U, 1)
```

```python
s = solution(G, 0)
s.MaxClique()
s.max
```
