```python
import os
import h5py
import numpy as np
from whale_gunshot_localization import config

ccb_data_root = os.path.join(*config['dataset']['ccb_data_directory'].split("/")[:-2],"single_examples")

idx1 = 44
idx2 = 49
idx3 = 53
```

### Combine Shorts

```python
# load
with h5py.File(os.path.join("/",ccb_data_root,f"gunshots_{idx1}.h5"), "r") as f:
    inputs_short1 = f["/data"][:]
    sensors_short1 = f["/sensors"][:]
    true_latlon_short1 = f["/location"][:]
    associations_short1 = f["/associations"][:]
    
with h5py.File(os.path.join("/",ccb_data_root,f"gunshots_{idx2}.h5"), "r") as f:
    inputs_short2 = f["/data"][:]
    sensors_short2 = f["/sensors"][:]
    true_latlon_short2 = f["/location"][:]
    associations_short2 = f["/associations"][:]
    
with h5py.File(os.path.join("/",ccb_data_root,f"gunshots_{idx3}.h5"), "r") as f:
    inputs_short3 = f["/data"][:]
    sensors_short3 = f["/sensors"][:]
    true_latlon_short3 = f["/location"][:]
    associations_short3 = f["/associations"][:]
    
inputs_short = np.concatenate((inputs_short1, inputs_short2, inputs_short3), axis=0)
sensors_short = np.concatenate((sensors_short1, sensors_short2, sensors_short3), axis=1)
true_latlon_short = np.concatenate((true_latlon_short1, true_latlon_short2, true_latlon_short3), axis=1)
associations_short = np.concatenate((np.zeros((1,17)), np.ones((1,17)), 2*np.ones((1,17))), axis=1)

with h5py.File(os.path.join("/",ccb_data_root,f"gunshots_{idx1}_{idx2}_{idx3}.h5"), "w") as f:
    f.create_dataset("/data", data=inputs_short, dtype=np.float32)
    f.create_dataset("/sensors", data=sensors_short, dtype=np.float32)
    f.create_dataset("/location", data=true_latlon_short, dtype=np.float32)
    f.create_dataset("/associations", data=associations_short, dtype=np.float32)
```

### Combine Longs

```python
# load
with h5py.File(os.path.join("/",ccb_data_root,f"gunshots_long_{idx1}.h5"), "r") as f:
    inputs_short1 = f["/data"][:]
    sensors_short1 = f["/sensors"][:]
    true_latlon_short1 = f["/location"][:]
    
with h5py.File(os.path.join("/",ccb_data_root,f"gunshots_long_{idx2}.h5"), "r") as f:
    inputs_short2 = f["/data"][:]
    sensors_short2 = f["/sensors"][:]
    true_latlon_short2 = f["/location"][:]
    
with h5py.File(os.path.join("/",ccb_data_root,f"gunshots_long_{idx3}.h5"), "r") as f:
    inputs_short3 = f["/data"][:]
    sensors_short3 = f["/sensors"][:]
    true_latlon_short3 = f["/location"][:]
    
inputs_short = np.concatenate((inputs_short1, inputs_short2, inputs_short3), axis=1)
sensors_short = np.concatenate((sensors_short1, sensors_short2, sensors_short3), axis=1)
true_latlon_short = np.concatenate((true_latlon_short1, true_latlon_short2, true_latlon_short3), axis=1)

with h5py.File(os.path.join("/",ccb_data_root,f"gunshots_long_{idx1}_{idx2}_{idx3}.h5"), "w") as f:
    f.create_dataset("/data", data=inputs_short, dtype=np.float32)
    f.create_dataset("/sensors", data=sensors_short, dtype=np.float32)
    f.create_dataset("/location", data=true_latlon_short, dtype=np.float32)
```
