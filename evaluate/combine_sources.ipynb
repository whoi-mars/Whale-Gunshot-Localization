```python
import os
import h5py
import numpy as np
from whale_gunshot_localization import config

ccb_data_root = os.path.join(*config['dataset']['ccb_data_directory'].split("/")[:-2],"single_examples")
idx = [49, 53]
```

### Combine Shorts

```python
file_name_save = os.path.join("/",ccb_data_root,"gunshots" + "".join([f"_{i}" for i in idx]) + ".h5")

for c, i in enumerate(idx):
    # load file
    with h5py.File(os.path.join("/",ccb_data_root,f"gunshots_{i}.h5"), "r") as f:
        inputs_short = f["/data"][:]
        sensors_short = f["/sensors"][:]
        true_latlon_short = f["/location"][:]
        associations_short = f["/associations"][:]
                              
    if c == 0:
        inputs_final = inputs_short
        sensors_final = sensors_short
        true_latlon_final = true_latlon_short
        associations_final = c * np.ones((1,inputs_short.shape[0]))
    else:
        inputs_final = np.concatenate((inputs_final, inputs_short), axis=0)
        sensors_final = np.concatenate((sensors_final, sensors_short), axis=1)
        true_latlon_final = np.concatenate((true_latlon_final, true_latlon_short), axis=1)
        associations_final = np.concatenate((associations_final, c * np.ones((1,inputs_short.shape[0]))), axis=1)
                              
with h5py.File(file_name_save, "w") as f:
    f.create_dataset("/data", data=inputs_final, dtype=np.float32)
    f.create_dataset("/sensors", data=sensors_final, dtype=np.float32)
    f.create_dataset("/location", data=true_latlon_final, dtype=np.float32)
    f.create_dataset("/associations", data=associations_final, dtype=np.float32)
```

### Combine Longs

```python
file_name_save = os.path.join("/",ccb_data_root,"gunshots_long" + "".join([f"_{i}" for i in idx]) + ".h5")

for c, i in enumerate(idx):
    with h5py.File(os.path.join("/",ccb_data_root,f"gunshots_long_{i}.h5"), "r") as f:
        inputs_short = f["/data"][:]
        sensors_short = f["/sensors"][:]
        true_latlon_short = f["/location"][:]
        
    if c == 0:
        inputs_final = inputs_short
        sensors_final = sensors_short
        true_latlon_final = true_latlon_short
        associations_final = associations_short
    else:
        inputs_final = np.concatenate((inputs_final, inputs_short), axis=1)
        sensors_final = np.concatenate((sensors_final, sensors_short), axis=1)
        true_latlon_final = np.concatenate((true_latlon_final, true_latlon_short), axis=1)
        
with h5py.File(file_name_save, "w") as f:
    f.create_dataset("/data", data=inputs_final, dtype=np.float32)
    f.create_dataset("/sensors", data=sensors_final, dtype=np.float32)
    f.create_dataset("/location", data=true_latlon_final, dtype=np.float32)
```
